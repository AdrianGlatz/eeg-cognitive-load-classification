# Cognitive-BCI Dataset Inspection

First look at the N-back EEG data, recording information, channels, and experimental events.

Subject 01, session 1 is used as a representative example for inspecting
the zero-, one-, and two-back recordings.

```python
import mne
import numpy as np
from pathlib import Path
```

```python
# 0-Back
zero_path = Path("..") / "data" / "sub-01" / "ses-S1" / "eeg" / "zeroBACK.set"

raw_zero = mne.io.read_raw_eeglab(zero_path, preload=True)

print("Duration:", raw_zero.times[-1])
print("Sampling rate:", raw_zero.info["sfreq"])
print("Channels:", len(raw_zero.ch_names))
print(raw_zero.get_channel_types(unique=True))

# Event occurence
event_names, counts = np.unique(
    raw_zero.annotations.description,
    return_counts=True
)

for name, count in zip(event_names, counts):
    print(name, ":", count)
```
