# Cognitive Load Classification

Training and evaluating a cognitive load classifier using subject-wise cross-validation.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path("..").resolve()
sys.path.insert(0, str(PROJECT_ROOT))

from src.dataset import build_dataset
from src.modeling import cross_validation_predictions, evaluate_predictions
```

```python
data_dir = PROJECT_ROOT / "data"

subject_names = [
    f"sub-{i:02d}"
    for i in range(1, 30)
]

X, y, groups = build_dataset(data_dir, subject_names)
```

```python
print(X.shape)
print(y.shape)
print(groups.shape)
```

```python
predictions = cross_validation_predictions(X, y, groups)
```

## Baseline Results

```python
accuracy, f1, matrix = evaluate_predictions(y, predictions)

print("Accuracy:", accuracy)
print("Macro F1:", f1)
print("Confusion Matrix:")
print(matrix)

low = matrix[0][0] / sum(matrix[0])
medium = matrix[1][1] / sum(matrix[1])
high = matrix[2][2] / sum(matrix[2])

print(f"Low: {low:.2%}")
print(f"Medium: {medium:.2%}")
print(f"High: {high:.2%}")
```
